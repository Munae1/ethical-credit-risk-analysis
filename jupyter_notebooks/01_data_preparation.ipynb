```python
#importing pandas 
import pandas as pd 
#load the cleaned dataset from project 2 
df = pd.read_csv("data/raw/credit_risk_dataset_cleaned.csv")
#show number of rows and columns 
print(df.shape)
df.head()
```

## Age Groups 

Age is a protected characteristic under the UK Equality Act 2010, so it is the main charactertic for this fairness assessment. With reference to the data, the average age is 27/28, with 75% of borrowers aged 30 or younger, meaning the larger part of the data is skewed young. 

```python
#Put group ages into bands for fairness audit 
df["age_group"] = pd.cut(
    df["person_age"],
    bins=[19, 24, 29, 39, 100],
    labels=["20-24", "25-29", "30-39", "40+"]
)
#count how many borrowers fall in each group
print(df["age_group"].value_counts().sort_index())

#show each group as a percentage of all borrowers 
print((df["age_group"].value_counts(normalize=True).sort_index() * 100).round(1))
```

## Observations 
- Majority of borrowers are of a younger age with 72% aged 20-29. 
- Borrowers 40+ only make 5.4% of the data (1756 people)
- This low amount of older borrowers can be considered as a fairness concern, as models using this data will learn primarily from younger borrowers, and will be less reliable for older ones. 

## Save the prepared dataset

```python
#save the prepared dataset
df.to_csv("data/processed/credit_risk_fairness.csv", index=False)

#confirm file is saved by checking shape
print(pd.read_csv("data/processed/credit_risk_fairness.csv").shape)

```
